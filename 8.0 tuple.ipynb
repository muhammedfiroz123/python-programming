#packing and unpacking

```python
t=("orange","mango","banana")
print(t)
```

```python
t=("orange","mango","banana")
(melon,kiwi,pappaya)=t
print(melon)
print(kiwi)
print(pappaya)
```

```python
t1=("orange","mango","banana","kiwi","cherry")
(melon,apple,*red)=t1
print(melon)
print(apple)
print(red)
```

```python
t1=("orange","mango","banana","kiwi","cherry")
(melon,*apple,red)=t1
print(melon)
print(apple)
print(red)
```

```python
## nested tuple
nst_tpl=((1,2,3),(45,6,7),('a','b','c'))
print(nst_tpl[2][2])
print(nst_tpl[1])
```

```python
## iteration tuple
for sub_tpl in nst_tpl:
  for item in sub_tpl:
    print(item,end=' ')
  print()
```
