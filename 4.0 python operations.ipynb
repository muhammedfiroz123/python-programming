## PYTHON OPERATIONS


```python
# arithmetic operation
a=10
b=5
add_res=a+b #addition
sub_res=a-b  #substraction
mult_res=a*b #multiplication
div_res=a/b #devision
floor_div_res=a//b  #floor devition
modules_res=a%b # modules operation
exponent_res=a**b # exponential

print(add_res)
print(sub_res)
print(div_res)
print(mult_res)
print(floor_div_res)
print(modules_res)
print(exponent_res)
```
