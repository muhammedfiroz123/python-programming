## custom exception

```python
class Error(Exception):
  pass

class dobException(Error):
  pass
```

```python
year=int(input('enter your dob'))
age=2025-year

try:
  if age<=30 and age>=20:
    print('this age i valid you can apply for this exam')

  else:
    raise dobException

except dobException:
  print('sorry ,your age should greater than 30 and less than 20')
```
