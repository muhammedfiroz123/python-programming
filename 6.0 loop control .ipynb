```python
## loop controll statement
## break

for i in range(10):
  if i==5:
    break
  print(i)


```

```python
## countinue

for i in range(10):
  if i%2==0:
    continue
  print(i)

```

```python
## pass
for i in range(5):
  if i==3:
    print('the number is ',i)
    pass
  print(i)

```

```python
## Nested loop
# loop inside a loop
print(range(3))
print('\n')
print(range(2))
for i in range(3):
  for j in range(2):
    print(f'i:{i}and j:{j}')
```
