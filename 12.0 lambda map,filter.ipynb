## MAP

```python
#mapping list and lambda
celcius=[34.5,56.3,45.5,34.9]
list(map(lambda x:float(9/5)*x +32 ,celcius))
```

```python
number=[1,2,3,4,5,6]
def square(x):
  return x*x
```

```python
number=[1,2,3,4,5,6]
list(map(square,number))
```

```python
a=[23,14,12,19]
b=[20,17,13,21]
c=[21,34,12,14]

list(map(lambda x,y:x*y,a,b))
```

```python
list(map(lambda x,y,z:x+y+z,a,b,c))
```

```python
lst=['1','2','3','4','5']
int_num=list(map(int,lst))
print(int_num)
print(type(int_num))
```

```python
def get_names(person):
  return person['name']


people=[
  {"name":'Firoz','age':23},{'name':'hashim','age':22}
]

list(map(get_names,people))
```

# FILTER


```python
fib=[0,1,1,4,5,3,7,9,12,24,26,45,55]
list(filter(lambda x: x%2,fib))
```

```python
list(filter(lambda x:x%2==0,fib))
```

```python
def age_greater_than_20(person):
  return person['age']>20


people=[
  {"name":'Firoz','age':23},{'name':'hashim','age':22},{'name':'rafi','age':20}
]

list(filter(age_greater_than_20,people))
```

```python
from functools import reduce
```

```python
fib=[0,1,1,4,5,3,7,9,12,24,26,45,55]
reduce(lambda x,y: x+y,fib)
```

```python
reduce(lambda a,b:a+b ,range(1,1023))
```

```python
f=lambda a,b: a if a<b else b #using reduce to compute minimum element from list
reduce(f,[47,15,26,789])
```
