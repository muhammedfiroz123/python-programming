```python
## function with multiple parameter

def number(x,y):
    return x+y
print(number(2,4))
print(number(3,4))
```

```python
def number(x,y,z=None):
    if z==None:
        print(x+y)
    else:
        print(x+y+z)
number(1,2)
number(3,4,5)
number(4,5)
    
```

```python
def number(a,b,c=None,flag=False):
    if flag==True:
        print("flag is true")
    if c==None:
        return a+b
    else:
        return a+b+c
print(number(1,2))
print(number(4,6,9))
print(number(1,3,4,flag=True))
```

```python
#positional and keyword argument
def print_numbers(*firoz):
  for number in firoz:
    print(number)

print_numbers(1,3,4,8,5,2,4,9,6,'firoz')
  
```

```python
#postional argumnet
def print_numbers(*args):
  for number in args:
    print(number)

print_numbers(1,3,4,8,5,2,4,9,6,'firoz')
```

```python
## keyword argumnet use ** 

def print_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(name='firoz',age=23,city='kollam')
```

```python
def print_details(*args,**kwargs):
  for val in args:
    print(f"positional argumnet:{val}")
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(1,4,6,3,7,'firoz',name='firoz',age=23,city='kollam')
```

# using function in another funcion


```python
def wage(w_hours):
    return w_hours*25

def bonus(w_hours):
    return wage(w_hours)+50
```

```python
wage(8),bonus(8)
```

## FUNCTION MORE EXAMPLES

```python
## TEMPERATURE CONVERTION
def convert_temperature(temp,unit):
  if unit=='C':
    return temp*9/5 + 32
  if unit=='F':
    return (temp-32) * 5/9
  else:
    return None
print(convert_temperature(34,'C'))
print(convert_temperature(99,"F"))
```
