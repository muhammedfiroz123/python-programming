### ENCAPSULATION

Using OOP in Python, we can restrict access to methods and variables.
This prevent data from direct modification which is called encapsulation.
In Python, we denote protected and private attribute using underscore as prefix i.e single “ _ “ or double “ __“.

```python
class computor:
    def __init__(self):
        self.__max_price=900
    def sell(self):
        print("selling in price:{}".format(self.__max_price))
    def sellMaxprice(self,price):
        self.__max_price=price
        
        
        
c=computor()
c.sell()
```

```python
c.__max_price=1000
c.sell()
```

```python
c.sellMaxprice(1000)
c.sell()
```
