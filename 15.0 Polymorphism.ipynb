## polymorphism


In programming, polymorphism means the same function name (but different signatures) being used for different types.
The key difference is the data types and number of arguments used in function.

```python
class bird:
    def intro(self):
        print("birds are fly")
    def flight(self):
        print("most of the birds are fly,but some!")
class sparrows(bird):
    def flight(self):
        print("sparrows are fly")

class ostrich(bird):
    def flight(self):
        print("ostrich can't fly")
        
objbird= bird()
objsparrow =sparrows()
objostrich=ostrich()

objbird.flight()
objbird.intro()
objsparrow.flight()
objsparrow.intro()
objostrich.intro()
        
```

```python
class Animal:
  def speak(self):
    return 'animal make some sounds'
  
class dog(Animal):
  def speak(self):
    return 'woof!'
  
class cat(Animal):
  def speak(self):
    return "meow!"
  

def animal_speak(animal):
  print(animal.speak())


Dog=dog()
print(Dog.speak())
Cat=cat()
print(f'cat make sound:- {Cat.speak()}')
animal_speak(Dog)
```

```python
class shape:
  def area(self):
    return 'the area of the figure'
  
class rectangle(shape):
  def __init__(self,width,height):
    self.width=width
    self.height=height

  def area(self):
    return self.width* self.height
  
class circle(shape):
  def __init__(self,radius):
    self.radius=radius

  def area (self):
    return 3.14* self.radius * self.radius
  

def print_area(shape):
  print(f'The area is= {shape.area()}')


Rectangle=rectangle(4,5)
circle1=circle(6)

print_area(Rectangle)
print_area(circle1)
    
```

### abstract methond in polymorphism

```python
from abc import ABC,abstractmethod

class vehicle(ABC):
  @abstractmethod
  def start_engine(self):
    pass

class car(vehicle):
  def start_engine(self):
    return 'the car engine started'
  
class bike(vehicle):
  def start_engine(self):
    return 'the bike engine started'
  
def start_vehicle(vehicle):
  print(vehicle.start_engine())


car1=car()
bike1=bike()
start_vehicle(bike1)
```
