## Magic methods

```python
class person():
  def __init__ (self,name,age):
    self.name=name
    self.age=age

  def __str__(self):
    return f"name is {self.name} and {self.age} years old"

  def __repr__(self):
    return f"person: name={self.name},age={self.age}"
## __repr__ is  magic method
## __str__ is the magic methond in this case lot of function used in some cases
person1=person('muhammed firoz',23)
print(person1)
print( repr(person1))
```
